Расшифровка [::-1]

Он переворачивает строку или список задом наперёд.

Скобки работают как [от : до]. Но на самом деле у среза есть третья, секретная часть: [от : до : шаг]

В коде [::-1] 
Первое число (откуда начать): Пусто. Значит, «бери с самого края».
Второе число (где закончить): Пусто. Значит, «иди до самого другого края».
Третье число (шаг): -1. Это самое главное! Обычный шаг — это 1 (идём вперёд по одному элементу). А шаг -1 означает «идём назад по одному элементу».

Как это связано с pandas (Анализом данных)?

Квадратные скобки — это главный способ достать из таблицы DataFrame столбец

Ты как бы говоришь pandas: "Открой таблицу и достань мне всё, что лежит в столбце 'Имя'"

repr(s) нужна для того, чтобы увидеть истинное лицо переменной. 
Она оборачивает текст в кавычки и показывает все скрытые символы

Цикл for: повторение действий для каждого элемента
Цикл for позволяет пройти по всем элементам последовательности: строки, списка, кортежа, множества или словаря.

Общая форма:

for элемент in последовательность:
    действие
Отступ внутри цикла обязателен. Именно отступ показывает Python, какие команды относятся к циклу.

Синтаксис такой: (итерационная переменная - это имя переменной цикла)

for итерационная_переменная in итерируемый объект:
    действие с итерируемым объектом

в качестве итерируемого объекта может быть список [] , строка "", функция, словарь {}, кортеж ()

```python
for amount in [1200, 2500, 1700, 3100]:
    print("Сумма заказа:", amount)

#здесь в качестве действия с итерируемым объектом представлено действие вывода на экран
```

5. Цикл while: повторение, пока условие истинно
Цикл while выполняется до тех пор, пока условие остается истинным.

Общая форма:

while условие:
    действие
Важно следить, чтобы условие когда-нибудь стало ложным. Иначе получится бесконечный цикл.

```python
i = 0

while i < 4:
    print("Текущее значение i:", i)
    i = i + 1
```

```python
temp = 20

while temp <= 29.3:
    print("Сейчас температура воды:", temp)
    temp = temp + 0.5
```

5.1. break: досрочная остановка цикла
Команда break прекращает выполнение цикла.

Это удобно, когда мы ищем нужное значение и хотим остановиться сразу после нахождения.

```python
amounts = [900, 1200, 2500, 7000, 1800]

for amount in amounts:
    print("Проверяем заказ:", amount)
    if amount > 5000:
        print("Найден крупный заказ:", amount)
        break
```

Чтобы ноутбук можно было запускать целиком, здесь используем заранее заданный список чисел. Ноль означает остановку обработки.

```python
numbers_from_user = [2, 3, 2, 5, 0, 10]
unique_squares = set()

for number in numbers_from_user:
    if number == 0:
        break
    unique_squares.add(number ** 2)

print("Введенные числа до остановки:", numbers_from_user)
print("Уникальные квадраты:", unique_squares)
```

```python
n = 5
numbers = []

for i in range(1, n + 1):
    numbers.append(i)
    print(*numbers, sep="")

#  Ключевые моменты этого кода:
#  append() — постепенно наполняет список
#  * (звездочка) — распаковывает список на отдельные элементы
#  sep="" — убирает пробелы между элементами при выводе
#  Лесенка получается, потому что внутри цмкла есть print
```

```python
n = 5
numbers = []

for i in range(1, n + 1):
    numbers.append(i)
    print(" " * (n - i), end="")
    print(*numbers, sep="", end="")
    print(*numbers[-2::-1], sep="")
```

9. Итоговая мини-практика: простая обработка заказов
Теперь объединим строки, списки, множества, словари и циклы.

У нас есть список заказов. Каждый заказ представлен словарем. Нужно:

вывести все заказы в понятном виде;
посчитать общую выручку;
найти уникальные города;
посчитать количество заказов со статусом оплачен;
собрать суммы продаж по городам.

```python
orders = [
    {"id": 1001, "city": "Москва", "amount": 2500, "status": "оплачен"},
    {"id": 1002, "city": "Казань", "amount": 1800, "status": "новый"},
    {"id": 1003, "city": "Москва", "amount": 4200, "status": "оплачен"},
    {"id": 1004, "city": "Самара", "amount": 900, "status": "отменен"},
    {"id": 1005, "city": "Казань", "amount": 3100, "status": "оплачен"},
]

print("Список заказов:")
for order in orders:
    print(f"Заказ {order['id']}: {order['city']}, {order['amount']} руб., статус: {order['status']}")

total_revenue = 0
unique_cities = set()
paid_orders_count = 0
revenue_by_city = {}

for order in orders:
    total_revenue = total_revenue + order["amount"]
    unique_cities.add(order["city"])
    
    if order["status"] == "оплачен":
        paid_orders_count = paid_orders_count + 1
    
    city = order["city"]
    amount = order["amount"]
    
    if city not in revenue_by_city:
        revenue_by_city[city] = 0
    revenue_by_city[city] = revenue_by_city[city] + amount

print("Общая выручка:", total_revenue)
print("Уникальные города:", unique_cities)
print("Количество оплаченных заказов:", paid_orders_count)
print("Выручка по городам:", revenue_by_city)
```
