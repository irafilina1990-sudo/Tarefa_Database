```python
students_info = {
    'name': 'Alice',
    'age': 22,
    'cource': 'DA'
}
print('исходный словарь: \n', students_info)
```

```python
students_data = {
    'name': ['Alice', 'Bob', 'Charle'],
    'age': [22, 30, 40],
    'cource': ['DA', 'Computer Sience', 'Physics']
}
print(f"исходный словарь: \n {students_data}" )
```

```python
students_df = pd.DataFrame(students_data)
```
